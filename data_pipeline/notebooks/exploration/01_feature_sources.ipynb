```python
import earthaccess
import xarray as xr
import os

earthaccess.login(persist=True)

results = earthaccess.search_data(
    concept_id="C2916514952-POCLOUD", # CCMP Winds
    temporal=(bbox["start_datetime"], bbox["end_datetime"])
)

raw_dir = os.path.join(bbox["output_directory"], "raw_jan_winds")
os.makedirs(raw_dir, exist_ok=True)

print(f"Downloading {len(results)} files to disk...")
earthaccess.download(results, local_path=raw_dir)

print("\nFiles downloaded! Processing local tensors...")
with xr.open_mfdataset(f"{raw_dir}/*.nc", combine='by_coords') as ds:
    # Crop geographically using your bbox variables
    ds_subset = ds.sel(
        latitude=slice(bbox["minimum_latitude"], bbox["maximum_latitude"]),
        longitude=slice(bbox["minimum_longitude"], bbox["maximum_longitude"])
    )
    
    # Isolate only the U and V wind vectors
    ds_subset = ds_subset[["uwnd", "vwnd"]]
    
    ds_subset = ds_subset.resample(time="1D").mean()
    
    # 5. Save the final merged and standardized file
    output_path = os.path.join(bbox["output_directory"], "04_jan_winds.nc")
    ds_subset.to_netcdf(output_path)

print(f"✅ Download and processing complete! File saved to: {output_path}")
```

```python
import earthaccess
import xarray as xr
import os

earthaccess.login(persist=True)

results = earthaccess.search_data(
    concept_id="C2098858642-POCLOUD", # OSCAR L4 Currents
    temporal=(bbox["start_datetime"], bbox["end_datetime"])
)

raw_dir = os.path.join(bbox["output_directory"], "raw_jan_currents")
os.makedirs(raw_dir, exist_ok=True)

print(f"Downloading {len(results)} files to disk...")
earthaccess.download(results, local_path=raw_dir)

print("\nFiles downloaded! Processing local tensors...")
with xr.open_mfdataset(f"{raw_dir}/*.nc", combine='by_coords') as ds:
    
    # Rename the coordinate *variables* (lat/lon) to match the *dimensions* (latitude/longitude)
    # This automatically promotes them to "Index Coordinates" so .sel() works!
    ds = ds.rename_vars({"lat": "latitude", "lon": "longitude"})

    # Crop geographically using your bbox variables
    ds_subset = ds.sel(
        latitude=slice(bbox["minimum_latitude"], bbox["maximum_latitude"]),
        longitude=slice(bbox["minimum_longitude"], bbox["maximum_longitude"])
    )
    
    # Isolate only the U and V total surface current vectors
    ds_subset = ds_subset[["u", "v"]]
    
    # 5. Save the final merged and standardized file
    output_path = os.path.join(bbox["output_directory"], "05_jan_currents.nc")
    ds_subset.to_netcdf(output_path)

print(f"Download and processing complete! File saved to: {output_path}")
```

```python
import xarray as xr

ds1 = xr.open_dataset("data_pipeline/examples/inputs_demo/01_jan_sst_.nc") 
ds1
```

```python
import matplotlib.pyplot as plt
import math

# 1. Automatically filter the dataset for even calendar days (e.g., 2, 4, 6...)
even_ds = ds1.isel(time=(ds1['time'].dt.day % 2 == 0))
total_plots = len(even_ds['time'])

# 2. Setup a dynamic grid (Creates a 3x5 grid for ~15 even days in a month)
cols = 5
rows = math.ceil(total_plots / cols)
fig, axes = plt.subplots(nrows=rows, ncols=cols, figsize=(24, 4 * rows), sharex=True, sharey=True)
axes = axes.flatten()

# 3. Loop through and plot each even day
for i in range(total_plots):
    ax = axes[i]
    
    # Extract data and convert Kelvin to Celsius!
    daily_slice = even_ds['analysed_sst'].isel(time=i) - 273.15
    
    # Extract a clean string of the date (YYYY-MM-DD) for the title
    date_str = str(even_ds['time'].isel(time=i).values)[:10]
    
    # Plot the slice
    daily_slice.plot(
        ax=ax,
        cmap='inferno',
        cbar_kwargs={'shrink': 0.8, 'label': 'Temp (°C)'}
    )
    
    # Formatting
    ax.set_title(f"{date_str}", fontsize=12, fontweight='bold')
    ax.set_xlabel("Longitude" if i >= (rows - 1) * cols else "")
    ax.set_ylabel("Latitude" if i % cols == 0 else "")

# 4. Hide any leftover empty subplots if the grid isn't perfectly filled
for j in range(total_plots, len(axes)):
    axes[j].axis('off')

plt.suptitle("Analysed Sea Surface Temperature - Even Days", fontsize=20, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
import xarray as xr
ds4 = xr.open_dataset("data_pipeline/examples/inputs_demo/04_jan_winds.nc")
ds4
```

```python
import matplotlib.pyplot as plt

# 1. Isolate EVEN days (Index 1 is Jan 2nd, Index 3 is Jan 4th, etc.)
ds_even_days = ds4.isel(time=slice(1, None, 2))

print(f"Plotting {len(ds_even_days.time)} even days for Zonal (U) and Meridional (V) winds...")

# 2. Plot U-Wind (Zonal: East-West)
ds_even_days.uwnd.plot(
    col="time",             
    col_wrap=4,             
    figsize=(18, 12),       
    cmap="PuOr_r",          # Purple-Orange colormap to show East vs West flow
    cbar_kwargs={"label": "Zonal Wind Velocity (m/s)"}
)
plt.suptitle("Zonal Wind (U-component) - Even Days of January 2021", y=1.02, fontsize=18, fontweight="bold")
plt.show()

# 3. Plot V-Wind (Meridional: North-South)
ds_even_days.vwnd.plot(
    col="time",             
    col_wrap=4,             
    figsize=(18, 12),       
    cmap="PiYG",            # Pink-Green colormap to show North vs South flow
    cbar_kwargs={"label": "Meridional Wind Velocity (m/s)"}
)
plt.suptitle("Meridional Wind (V-component) - Even Days of January 2021", y=1.02, fontsize=18, fontweight="bold")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Isolate ODD days (Index 0 is Jan 1st, Index 2 is Jan 3rd, etc.)
ds_odd_days = ds4.isel(time=slice(0, None, 2))

print(f"Plotting {len(ds_odd_days.time)} odd days for combined Wind Speed...")

# 2. Calculate the mixed UV wind speed (magnitude) mathematically
wind_speed = np.sqrt(ds_odd_days.uwnd**2 + ds_odd_days.vwnd**2)
wind_speed.name = "Wind Speed (m/s)" # Names the colorbar automatically

# 3. Plot the combined mixed energy
wind_speed.plot(
    col="time",             
    col_wrap=4,             
    figsize=(18, 12),       
    cmap="viridis",         # Perceptually uniform colormap for magnitude intensity
    cbar_kwargs={"label": "Total Wind Speed Magnitude (m/s)"}
)
plt.suptitle("Combined Wind Speed (UV Mixed) - Odd Days of January 2021", y=1.02, fontsize=18, fontweight="bold")
plt.show()
```
