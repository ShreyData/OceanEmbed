```python
import numpy as np
import matplotlib.pyplot as plt

var_name = 'analysed_sst' 
ds_target = sst_sample

total_nulls = ds_target[var_name].isnull().sum().compute().item()
total_zeros = (ds_target[var_name] == 0.0).sum().compute().item()

print(f"Audit for '{var_name}':")
print(f"  -> Total Missing (NaN): {total_nulls}")
print(f"  -> Total Exact Zeros: {total_zeros}")

nulls_per_pixel = ds_target[var_name].isnull().sum(dim='time').compute()

total_land_pixels = (nulls_per_pixel == 31).sum().item()
print(f"  -> Permanent Land Pixels (100% NaN): {total_land_pixels} out of {100 * 240} total pixels")

plt.figure(figsize=(10, 4))
nulls_per_pixel.plot(cmap='magma')
plt.title(f"Null Distribution Map for {var_name}")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Define the datasets and variables to audit
variables_to_audit = [
    ("SST", sst_sample, 'analysed_sst'),
    ("SSS", sss_sample, 'sos'),
    ("SSH", ssh_sample, 'sla'),
    ("Winds (U-comp)", winds_sample, 'uwnd'),
    ("Winds (V-comp)", winds_sample, 'vwnd'),
    ("Currents (U-comp)", currents_sample, 'u'),
    ("Currents (V-comp)", currents_sample, 'v')
]

fig, axes = plt.subplots(2, 4, figsize=(22, 10))
axes = axes.flatten()

print("=" * 60)
print("DATASET AUDIT & NULL EXPLORATION")
print("=" * 60)

for i, (name, ds, var) in enumerate(variables_to_audit):
    # 1. Compute occurrences of strictly NaNs across the time dimension
    null_count = ds[var].isnull().sum(dim='time').compute()
    
    # 2. Compute exact global metrics
    total_nulls = null_count.sum().item()
    
    time_steps = ds.sizes['time'] 
    total_pixels = ds.sizes['latitude'] * ds.sizes['longitude']
    
    # Permanent land (pixels where every single time step is NaN)
    permanent_land = (null_count == time_steps).sum().item()
    
    # 3. Print the text audit
    print(f"Audit for '{name}' ({var}):")
    print(f"  -> Total Missing (NaN): {total_nulls}")
    print(f"  -> Permanent Masked Pixels (100% NaN): {permanent_land} out of {total_pixels} pixels\n")
    
    # 4. Plot the heatmap using only the null_count
    null_count.plot(
        ax=axes[i], 
        cmap='magma', 
        add_colorbar=True, 
        vmin=0, 
        vmax=time_steps
    )
    axes[i].set_title(f"{name} (NaNs Only)")
    axes[i].set_xlabel("Longitude")
    axes[i].set_ylabel("Latitude")

# Hide the unused 8th subplot
axes[-1].axis('off')

plt.tight_layout()
plt.show()
```
