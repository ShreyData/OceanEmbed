```python
import matplotlib.pyplot as plt

# Define the datasets and variables to audit
variables_to_audit = [
    ("SST", sst, 'analysed_sst'),
    ("SSS", sss, 'sos'),
    ("SSH", ssh, 'sla'),
    ("Winds (U-comp)", winds, 'uwnd'),
    ("Winds (V-comp)", winds, 'vwnd'),
    ("Currents (U-comp)", currents, 'u'),
    ("Currents (V-comp)", currents, 'v')
]

fig, axes = plt.subplots(2, 4, figsize=(22, 10))
axes = axes.flatten()

print("=" * 60)
print("DATASET AUDIT & NULL EXPLORATION")
print("=" * 60)

for i, (name, ds, var) in enumerate(variables_to_audit):
    # 1. Compute occurrences of strictly NaNs across the time dimension
    null_count = ds[var].isnull().sum(dim='time').compute()
    
    # 2. Compute exact global metrics
    total_nulls = null_count.sum().item()
    
    time_steps = ds.sizes['time'] 
    total_pixels = ds.sizes['latitude'] * ds.sizes['longitude']
    
    # Permanent land (pixels where every single time step is NaN)
    permanent_land = (null_count == time_steps).sum().item()
    
    # 3. Print the text audit
    print(f"Audit for '{name}' ({var}):")
    print(f"  -> Total Missing (NaN): {total_nulls}")
    print(f"  -> Permanent Masked Pixels (100% NaN): {permanent_land} out of {total_pixels} pixels\n")
    
    # 4. Plot the heatmap using only the null_count
    null_count.plot(
        ax=axes[i], 
        cmap='magma', 
        add_colorbar=True, 
        vmin=0, 
        vmax=time_steps
    )
    axes[i].set_title(f"{name} (NaNs Only)")
    axes[i].set_xlabel("Longitude")
    axes[i].set_ylabel("Latitude")

# Hide the unused 8th subplot
axes[-1].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Load the masked target dataset (adjust path if necessary)
ds_target = trgt

# Isolate the surface layer (depth=0) and compute null occurrences across time
var_name = 'thetao'
null_count = ds_target[var_name].isel(depth=0).isnull().sum(dim='time').compute()
time_steps = ds_target.sizes['time']
total_pixels = ds_target.sizes['latitude'] * ds_target.sizes['longitude']
permanent_land = (null_count == time_steps).sum().item()

print("=" * 60)
print(f"Audit for 'GLORYS Target' ({var_name}):")
print(f"  -> Permanent Masked Pixels (100% NaN): {permanent_land} out of {total_pixels} pixels")
print("=" * 60)

# Plot the heatmap
plt.figure(figsize=(7, 6))
null_count.plot(
    cmap='magma', 
    add_colorbar=True, 
    vmin=0, 
    vmax=time_steps
)
plt.title("GLORYS Target Data (Not Masked)")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.tight_layout()
plt.show()
```

```python
import xarray as xr
import pandas as pd


ds_target = trgt

time_steps = ds_target.sizes['time']
total_pixels = ds_target.sizes['latitude'] * ds_target.sizes['longitude']

print(f"Total time steps: {time_steps}")
print(f"Total pixels per slice: {total_pixels}")

# Calculate how many days each pixel is valid (non-NaN)
print("\nCalculating valid days per pixel...")
valid_days_per_pixel = ds_target['thetao'].isel(depth=0).notnull().sum(dim='time').compute()

# Count the frequency of each validity duration
pixel_counts = pd.Series(valid_days_per_pixel.values.flatten()).value_counts().sort_index(ascending=False)

print("\n" + "=" * 60)
print("PIXEL RELIABILITY AUDIT (Valid Days vs. Pixel Count)")
print("=" * 60)
print(f"{'Valid Days':<15} | {'Number of Pixels':<20}")
print("-" * 60)

# Print the top distribution
for valid_days, num_pixels in pixel_counts.head(25).items():
    if valid_days == time_steps:
        print(f"{valid_days:<15} | {num_pixels:<20} <-- PERFECT OCEAN")
    elif valid_days == 0:
        print(f"{valid_days:<15} | {num_pixels:<20} <-- PERMANENT LAND")
    else:
        print(f"{valid_days:<15} | {num_pixels:<20}")

# Print the permanent land at the bottom if it didn't fit in the head(25)
if 0 not in pixel_counts.head(25).index and 0 in pixel_counts.index:
    print("...")
    print(f"{0:<15} | {pixel_counts[0]:<20} <-- PERMANENT LAND")

print("-" * 60)
scattered_pixels = pixel_counts[(pixel_counts.index > 0) & (pixel_counts.index < time_steps)].sum()
print(f"Total scattered pixels (valid some days, NaN on others): {scattered_pixels}")
print("=" * 60)
```
